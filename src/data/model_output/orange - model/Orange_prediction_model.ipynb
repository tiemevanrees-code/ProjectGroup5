```python
import pandas as pd

df = pd.read_csv(
    'orange_power_forecast_2022_daily.csv')

df.head()
```

```python
df = df[['date', 'Random Forest']]
df = df.drop(index=[0, 1]).reset_index(drop=True)

df.head()
```

```python
df = df.rename(columns={'Random Forest': 'Orange Model - Predicted Wind Power (kWh)'})
df['Orange Model - Predicted Wind Power (kWh)'] = df['Orange Model - Predicted Wind Power (kWh)'].astype(float).astype(int)

df["Orange Model - Predicted Wind Power (kWh)"] = df["Orange Model - Predicted Wind Power (kWh)"] * 2


df.head(10)
```

```python
output_path = 'orange_power_forecast_2022_daily_clean.csv'
df.to_csv(output_path, index=False)

print(f'Saved {len(df)} rows to {output_path}')
```
