```python
import pandas as pd

df = pd.read_csv(r'wind-2022-uur-data.csv')

# extract just the date part from the 'validfrom' column
df['date'] = pd.to_datetime(df['validfrom (UTC)']).dt.date

df = df.rename(columns={"volume (kWh)": 'Actual Energy created (kWh)'})


# sum numeric columns per day, keep the date
df_daily = df.groupby('date').sum(numeric_only=True).reset_index()

# Drop first row (incomplete day)
df_daily = df_daily.iloc[1:]

# Keep only the columns we want
df_daily = df_daily[['date', 'Actual Energy created (kWh)']]

df_daily.to_csv("wind-2022-uur-data-reformated.csv", index=False)
print(f"Saved: {"wind-2022-uur-data-reformated.csv"} ({len(df_daily)} rows)")
```
